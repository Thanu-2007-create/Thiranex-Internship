```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
df = pd.read_csv("train.csv", encoding='latin1')
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True, errors='coerce')
```

```python
df['Order Date'].head()
df['Order Date'].isna().sum()
```

```python
monthly_sales = df.groupby(
    pd.Grouper(key='Order Date', freq='ME')
)['Sales'].sum()
```

```python
plt.figure(figsize=(10,5))
plt.plot(monthly_sales)
plt.title("Monthly Sales Trend")
plt.show()
```

```python
sales_df = monthly_sales.reset_index()
sales_df['Month_Number'] = range(len(sales_df))
```

```python
X = sales_df[['Month_Number']]
y = sales_df['Sales']

model = LinearRegression()
model.fit(X, y)
```

```python
sales_df['Predicted'] = model.predict(X)
```

```python
mae = mean_absolute_error(y, sales_df['Predicted'])
r2 = r2_score(y, sales_df['Predicted'])

print("MAE:", mae)
print("R2 Score:", r2)
```

```python
plt.figure(figsize=(10,5))
plt.plot(sales_df['Sales'], label="Actual")
plt.plot(sales_df['Predicted'], label="Predicted")
plt.legend()
plt.show()
```

```python
future = pd.DataFrame({
    'Month_Number': range(len(sales_df), len(sales_df)+6)
})

forecast = model.predict(future)
print(forecast)
```
